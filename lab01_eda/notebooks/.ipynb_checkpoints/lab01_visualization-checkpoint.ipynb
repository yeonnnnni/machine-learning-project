```python
import sys
sys.path.insert(0, '../src')
import matplotlib.pyplot as plt
import lab01

raw = lab01.load_data('../data/cafe_sales.csv')
clean = lab01.clean_data(raw)
raw.head()
```

## Plot 1 — Missing values per column (raw data)

```python
# ============ TODO (Plot 1) ============
# Bar chart of lab01.summarize_missing(raw).
# Label both axes and give the figure a title.
# =======================================
missing = lab01.summarize_missing(raw)
print(missing)

plt.figure(figsize = (6, 4))
plt.bar(missing.index, missing.values)

plt.xlabel("column")
plt.ylabel("count missing values")
plt.title("Missing values per column")

plt.tight_layout()
plt.savefig("plot1.png")
```

## Plot 2 — Distribution of `total_price` (clean data)

```python
# ============ TODO (Plot 2) ============
# Histogram of clean['total_price'] (~30 bins).
# Briefly note in a comment: is it skewed? why?
# Most total_price values are below 200,000, 
# while a few orders are around 350,000 and 950,000–1,000,000. 
# Therefore, the distribution is right-skewed.
# =======================================

# 'total_price' 열의 값을 30개 구간으로 나눠서 그리기
# x축은 'total_price', y축은 count

import matplotlib.ticker as ticker

plt.figure(figsize = (12, 4))

plt.hist(clean['total_price'], bins=30)

plt.xlabel("total_price")
plt.ylabel("count")
plt.title("Distribution of total_price")

plt.gca().xaxis.set_major_formatter(
    ticker.StrMethodFormatter("{x:,.0f}")
)

plt.tight_layout()
plt.savefig("plot2.png")
plt.show()

```

## Plot 3 — `customer_rating` by category (clean data)

```python
# ============ TODO (Plot 3) ============
# One boxplot per category on a single axis.
# Hint: clean.boxplot(column='customer_rating', by='category')
# =======================================

plt.figure(figsize = (6,4))
# category 별로 customer_rating 분포를 boxplot으로 그리고
clean.boxplot(column='customer_rating', by='category')

# 이 boxplot들을 하나의 axis에 두고
# x축은 category, y축은 customer rating
plt.xlabel("category")
plt.ylabel("customer_rating")
plt.title("customer_rating by category")
plt.suptitle("")

plt.tight_layout()
plt.savefig("plot2.png")
plt.show()

```
